**(13) What is the difference between for and while loops?**

Ans: for loop is used to iterate lists, tuples, sets, etc whereas while loop is used to repeat code until a specific condition is met.



**(14) Describe a scenario where a while loop is more suitable than a for loop?**

Ans: while loops are more suitable when the number of iterations is unknown and depends on a conditions, such as reading data from a file until a specific line is found.
